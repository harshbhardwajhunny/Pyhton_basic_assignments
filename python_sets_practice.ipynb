### Q5
Given:
```python
A = {2, 4, 6, 8}
B = {4, 8, 12}
```
Find Union, Intersection, A-B

````python
A = {2, 4, 6, 8}
B = {4, 8, 12}
print("union :",A.union(B))
print("intersection:",A.intersection(B))
print("difference:",A.difference(B))
````

### Q6
Check whether 15 exists in the set {5, 10, 15, 20}.

````python
my_set={5,10,15,20}
for i in my_set:
  if i==15:
    print("15 exists in the set")
    break
else:
    print("not found ")
````
